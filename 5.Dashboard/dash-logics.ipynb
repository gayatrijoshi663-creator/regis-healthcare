#### 5. Incident Count by Severity

```python
%sql
select d.severity,
       count("*") as Incident_Count
         from incidents f 
         left join incident_type d 
on f.incident_type_key = d.incident_type_key 
group by severity order by Incident_Count ;
```

#### Facility-wise Incident Count

```python
%sql
select d.facility_name,
  count(*) as  incident_count 
  from incidents f join facilities d 
  on f.facility_key = d.facility_key  
  join incident_type i
  on f.incident_type_key = i.incident_type_key 
group by d.facility_name
       ORDER BY
    incident_count DESC;
```

#### Incident Count Monthly Trend

```python
%sql 
select 
   d.year,
   d.month_name,
   j.severity,
   count(*) as incident_count
   FROM incidents i
JOIN date d
ON i.incident_date_key = d.date_key
join incident_type j
on j.incident_type_key = i.incident_type_key
GROUP BY
    d.year,
    d.month_name,
    j.severity

ORDER BY
    d.year,
    d.month_name,
    j.severity;
```

#### Critical Incidents Only

```python
%sql
SELECT
   t.severity,
    COUNT(*) AS critical_incidents
FROM incidents i
join incident_type t
on t.incident_type_key = i.incident_type_key
group by t.severity having severity='Critical';
```
